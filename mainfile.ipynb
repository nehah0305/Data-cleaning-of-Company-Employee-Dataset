```python
import pandas as pd
import numpy as np
```

```python
FILE_NAME = "messy_company_employee_dataset.csv"

df = pd.read_csv(FILE_NAME)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
df_original = df.copy()
```

```python
print("\n========== FIRST 5 ROWS ==========")
display(df.head())

print("\n========== LAST 5 ROWS ==========")
display(df.tail())

print("\n========== RANDOM SAMPLE ==========")
display(df.sample(min(5, len(df))))

print("\n========== DATASET SHAPE ==========")
print(df.shape)

print("\n========== COLUMN NAMES ==========")
print(df.columns.tolist())

print("\n========== DATA TYPES ==========")
print(df.dtypes)

print("\n========== DATASET INFO ==========")
df.info()

print("\n========== DESCRIPTIVE STATISTICS ==========")
display(df.describe(include="all"))
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
)

print("\nCleaned column names:")
print(df.columns.tolist())
```

```python
print("\n========== MISSING VALUES BEFORE CLEANING ==========")

missing_before = df.isnull().sum()

missing_percentage_before = (
    df.isnull().sum() / len(df)
) * 100

missing_report = pd.DataFrame({
    "Missing_Values": missing_before,
    "Percentage": missing_percentage_before
})

display(missing_report)
```

```python
print("\n========== DUPLICATES BEFORE CLEANING ==========")

duplicate_count_before = df.duplicated().sum()

print("Duplicate rows:", duplicate_count_before)

if duplicate_count_before > 0:
    display(df[df.duplicated(keep=False)])
```

```python
print("\n========== UNIQUE CATEGORICAL VALUES ==========")

categorical_columns = [
    "Department",
    "Job_Title",
    "Gender",
    "City",
    "Work_Mode"
]

for column in categorical_columns:

    if column in df.columns:

        print(f"\n--- {column} ---")
        print(df[column].value_counts(dropna=False))
```

```python
before = len(df)

df = df.dropna(how="all")

after = len(df)

print(
    "\nCompletely empty rows removed:",
    before - after
)
```

```python
before = len(df)

df = df.drop_duplicates()

after = len(df)

print(
    "Duplicate rows removed:",
    before - after
)
```

```python
text_columns = df.select_dtypes(
    include="object"
).columns

for column in text_columns:
    df[column] = df[column].str.strip()

print("\nExtra spaces removed from text columns.")
```

```python
if "Gender" in df.columns:

    df["Gender"] = (
        df["Gender"]
        .astype("string")
        .str.strip()
        .str.lower()
    )

    df["Gender"] = df["Gender"].replace({
        "m": "Male",
        "male": "Male",
        "man": "Male",

        "f": "Female",
        "female": "Female",
        "woman": "Female"
    })
```

```python
if "Department" in df.columns:

    df["Department"] = (
        df["Department"]
        .astype("string")
        .str.strip()
        .str.lower()
    )

    df["Department"] = df["Department"].replace({

        "hr": "Human Resources",
        "human resource": "Human Resources",
        "human resources": "Human Resources",

        "it": "Information Technology",
        "information tech": "Information Technology",
        "information technology": "Information Technology",

        "fin": "Finance",
        "finance": "Finance",

        "marketing": "Marketing",
        "mkt": "Marketing",

        "sales": "Sales"
    })
```

```python
if "Work_Mode" in df.columns:

    df["Work_Mode"] = (
        df["Work_Mode"]
        .astype("string")
        .str.strip()
        .str.lower()
    )

    df["Work_Mode"] = df["Work_Mode"].replace({

        "office": "On-site",
        "on site": "On-site",
        "onsite": "On-site",

        "remote": "Remote",
        "work from home": "Remote",
        "wfh": "Remote",

        "hybrid": "Hybrid"
    })
```

```python
numeric_columns = [
    "Age",
    "Annual_Salary",
    "Experience_Years",
    "Performance_Score"
]

for column in numeric_columns:

    if column in df.columns:

        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )

print("\nNumeric columns converted.")
```

```python
if "Joining_Date" in df.columns:

    df["Joining_Date"] = pd.to_datetime(
        df["Joining_Date"],
        errors="coerce"
    )

print("Joining_Date converted to datetime.")
```

```python
if "Age" in df.columns:

    invalid_age = (
        (df["Age"] < 18) |
        (df["Age"] > 100)
    )

    print(
        "\nInvalid age records:",
        invalid_age.sum()
    )

    df.loc[invalid_age, "Age"] = np.nan
```

```python
if "Experience_Years" in df.columns:

    invalid_experience = (
        df["Experience_Years"] < 0
    )

    print(
        "Invalid experience records:",
        invalid_experience.sum()
    )

    df.loc[
        invalid_experience,
        "Experience_Years"
    ] = np.nan
```

```python
if "Performance_Score" in df.columns:

    invalid_score = (
        (df["Performance_Score"] < 0) |
        (df["Performance_Score"] > 10)
    )

    print(
        "Invalid performance scores:",
        invalid_score.sum()
    )

    df.loc[
        invalid_score,
        "Performance_Score"
    ] = np.nan
```

```python
if "Annual_Salary" in df.columns:

    invalid_salary = (
        df["Annual_Salary"] <= 0
    )

    print(
        "Invalid salary records:",
        invalid_salary.sum()
    )

    df.loc[
        invalid_salary,
        "Annual_Salary"
    ] = np.nan
```

```python
numeric_columns = [
    "Age",
    "Annual_Salary",
    "Experience_Years",
    "Performance_Score"
]

print("\n========== NUMERICAL IMPUTATION ==========")

for column in numeric_columns:

    if column in df.columns:

        missing = df[column].isnull().sum()

        if missing > 0:

            median_value = df[column].median()

            df[column] = df[column].fillna(
                median_value
            )

            print(
                f"{column}: {missing} missing values "
                f"filled with median = {median_value}"
            )
```

```python
categorical_columns = [
    "Department",
    "Job_Title",
    "Gender",
    "City",
    "Work_Mode"
]

print("\n========== CATEGORICAL IMPUTATION ==========")

for column in categorical_columns:

    if column in df.columns:

        missing = df[column].isnull().sum()

        if missing > 0:

            mode_value = df[column].mode()[0]

            df[column] = df[column].fillna(
                mode_value
            )

            print(
                f"{column}: {missing} missing values "
                f"filled with mode = {mode_value}"
            )
```

```python
if "Employee_Name" in df.columns:

    missing_names = df["Employee_Name"].isnull().sum()

    if missing_names > 0:

        df["Employee_Name"] = df[
            "Employee_Name"
        ].fillna("Unknown")

        print(
            f"Employee_Name: {missing_names} "
            "missing values replaced with 'Unknown'"
        )
```

```python
if "Joining_Date" in df.columns:

    missing_dates = df["Joining_Date"].isnull().sum()

    print(
        "\nMissing Joining_Date values:",
        missing_dates
    )
```

```python
if "Employee_ID" in df.columns:

    duplicate_ids = df[
        "Employee_ID"
    ].duplicated().sum()

    print(
        "\nDuplicate Employee_IDs:",
        duplicate_ids
    )

    if duplicate_ids > 0:

        display(
            df[
                df["Employee_ID"].duplicated(
                    keep=False
                )
            ].sort_values("Employee_ID")
        )
```

```python
if "Employee_ID" in df.columns:

    before = len(df)

    df = df.drop_duplicates(
        subset=["Employee_ID"],
        keep="first"
    )

    after = len(df)

    print(
        "Duplicate Employee_ID records removed:",
        before - after
    )
```

```python
print("\n========== MISSING VALUES AFTER CLEANING ==========")

missing_after = df.isnull().sum()

missing_percentage_after = (
    df.isnull().sum() / len(df)
) * 100

final_missing_report = pd.DataFrame({

    "Missing_Values": missing_after,

    "Percentage": missing_percentage_after

})

display(final_missing_report)
```

```python
print("\n========== DUPLICATES AFTER CLEANING ==========")

print(
    "Duplicate rows:",
    df.duplicated().sum()
)
```

```python
comparison = pd.DataFrame({

    "Metric": [
        "Rows",
        "Columns",
        "Missing Values",
        "Duplicate Rows"
    ],

    "Before Cleaning": [

        df_original.shape[0],

        df_original.shape[1],

        df_original.isnull().sum().sum(),

        df_original.duplicated().sum()
    ],

    "After Cleaning": [

        df.shape[0],

        df.shape[1],

        df.isnull().sum().sum(),

        df.duplicated().sum()
    ]
})

print("\n========== BEFORE VS AFTER ==========")

display(comparison)
```

```python
print("\n========== FINAL DATASET ==========")

print("Shape:", df.shape)

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 rows:")
display(df.head())

print("\nDescriptive statistics:")
display(df.describe(include="all"))
```

```python
OUTPUT_FILE = "cleaned_company_employee_dataset.csv"

df.to_csv(
    OUTPUT_FILE,
    index=False
)

print(
    f"\nCleaned dataset saved successfully as: "
    f"{OUTPUT_FILE}"
)
```
